```python
import pandas as pd


# Sample data
data = {
    'Name': ['Cust_num', 'Drone_num', 'Segment_num', "Result"],
}
count = 0
text = ""
with open("instances.txt", "r") as f:
    text = f.read()
raw_data = text.split()
for d in raw_data:
    id = "row {row}".format(row=count)
    data[id] = d.split("_")
    count += 1
print(data)
df = pd.DataFrame(data).T

# Save to Excel
df.to_excel('output.xlsx', index=False)
```
